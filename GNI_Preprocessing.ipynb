## HTML tag removing
Pattern to remove HTML tag = '<.*?>'

Pattern to remove url = 'https?://\S+|www\.\S+'

```python
import re
def remove_html(texts):
    clear_text = list()
    for text in texts:
        pattern = '<.*?>'
        clear_text.append(re.sub(pattern,'',text))
    return clear_text
```

```python
df['review'] = remove_html(df['review'])
df['review'][0]
```

### Remove URL

```python
def remove_url(texts):
    clear_text = list()
    for text in texts:
        pattern = 'https?://\S+|www\.\S+'
        clear_text.append(re.sub(pattern,'',text))
    return clear_text
```

```python
df['review'] = remove_url(df['review'])
df['review'][0]
```

## Punctuation handling
As puntuation marks have no value or meaning to computer we remove them.

```python
import string
```

```python
def remove_punc(text):
    chars = string.punctuation
    for char in chars:
        text = text.replace(char,' ')
    return text
```

```python
text = "af,. gag., ghl//.."
remove_punc(text)
```

```python
df['review'] = df['review'].apply(remove_punc)
df['review'][0]
```

## Chat conversation handling
Utilize short forms from texts.

```python
chat_words = {
    "u": "you",
    "brb": "be right back",
    "tbh": "to be honest",
    "idk": "I don’t know",
    "omw": "on my way",
    "afk": "away from keyboard",
    "np": "no problem",
    "thx": "thanks",
    "btw": "by the way",
    "imo": "in my opinion",
    "lol": "laugh out loud",
    "omg": "oh my god",
    "rofl": "rolling on the floor laughing",
    "icymi": "in case you missed it"
}
```

```python
def replace_chatwords(texts):
    new_text = list()
    for text in texts.split():
        if text in chat_words:
            new_text.append(chat_words[text])
        else:
            new_text.append(text)
    return " ".join(new_text)
```

```python
replace_chatwords('Do this work icymi')
```

```python
df['review'] = df['review'].apply(replace_chatwords)
df['review'][0]
```

## Stop words

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = list()
    for word in text.split():
        if  word in stopwords.words('english'):
            new_text.append(' ')
        else:
            new_text.append(word)
    return ' '.join(new_text)
```

```python
remove_stopwords('one of the other reviewers has mentioned that after watching just 1 oz episode')
```

```python
df['review'] = df['review'].apply(remove_stopwords)
df['review'][0]
```

## Remove Emoji

```python
def remove_emoji(text):
    def remove_emoji(text):
        emoji_pattern = re.compile(
            "["
            u"\U0001F600-\U0001F64F"  # emoticons
            u"\U0001F300-\U0001F5FF"  # symbols & pictographs
            u"\U0001F680-\U0001F6FF"  # transport & map symbols
            u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
            "]+",
            flags=re.UNICODE
        )
        return re.sub(emoji_pattern,'', text)
```

```python
remove_emoji('I am ')
```

## Stemmer
Transform verbs to root form. But it changes many words. They are not human readeble.

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return ' '.join([ps.stem(word) for word in text.split()])
```

```python
text = "walk walks eating walked"
stem_words(text)
```

### Lemmatization
Human readeble output

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
wordnet_lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize(text):
    return ' '.join([wordnet_lemmatizer.lemmatize(word) for word in text.split()])
```

```python
lemmatize(text)
```
