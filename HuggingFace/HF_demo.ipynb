````python
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')

text = 'I was not happy with that movie'

# Tokenize the text
tokens = tokenizer.tokenize(text)
print(f"Tokens : {tokens}")
````

````python
# Converts tokens to Inputs IDs(Conver tokens to identical number)
input_ids = tokenizer.convert_tokens_to_ids(tokens)
print(f"Input IDs : {input_ids}")
````

By default Hugging Face Represents sentence:


```
<sos>'I was not happy with that movie'<eos>
```



Encoded encodes sos with 101 and eos with 102.

````python
# Convert numbers into vectors
encoded_input = tokenizer(text)
print(f"Encoded input : {encoded_input}")
````

````python
# Decode token
decoded_output = tokenizer.decode(input_ids)
print(f"Decoded Output : {decoded_output}")
````

# Finetuning LM with custom data
1. Install library
2. Load data
3. Preprocess data


## 2. Load dataset

````python
!pip install datasets
````

````python
from datasets import load_dataset
dataset = load_dataset('stanfordnlp/imdb')
````

````python
dataset
````

````python
dataset['train'][0]
````

## 2. Preprocess data

````python
from transformers import AutoTokenizer
````

````python
# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')

# Tokenize the dataset
def tokenizer_function(data):
  return tokenizer(data['text'],padding = 'max_length',truncation = True)

tokenized_dataset = dataset.map(tokenizer_function,batched = True)
````

````python
tokenized_dataset
````

````python
tokenized_dataset['train'][0]
````

## Step4: Set up Training Arguments

````python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir = './result',
    eval_strategy = 'epoch',
    learning_rate = 2e-5,
    per_device_train_batch_size = 16,
    per_device_eval_batch_size = 16,
    num_train_epochs = 1,
    weight_decay = 0.01
)
training_args
````

## Step 5 ;Initialize model

````python
from transformers import AutoModelForSequenceClassification,Trainer

# Load the pre-trained model
model = AutoModelForSequenceClassification.from_pretrained('bert-base-uncased',num_labels = 2)
trainer = Trainer(
    model = model,
    args = training_args,
    train_dataset = tokenized_dataset['train'],
    eval_dataset = tokenized_dataset['test']
)
````

````python
# Train the model
trainer.train()
````

## Step7 : Evaluate the model

````python
results = trainer.evaluate()
print(results)
````
